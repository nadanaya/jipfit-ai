# 집핏 AI 합성 데이터 EDA

이 노트북은 합성 데이터 품질과 클래스 분포를 빠르게 확인합니다. 실제 개인 금융 데이터가 아닙니다.

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
path = ROOT / 'data' / 'processed' / 'synthetic_housing_scenarios.csv'
df = pd.read_csv(path)
df.shape, df.head()
```

```python
quality = pd.DataFrame({
    'missing': df.isna().sum(),
    'dtype': df.dtypes.astype(str),
    'unique': df.nunique(),
})
quality
```

```python
class_share = (df['burden_class'].value_counts(normalize=True).sort_index() * 100).round(2)
class_share.rename(index={0: '안정', 1: '주의', 2: '위험'})
```

```python
df.groupby('burden_class')[['monthly_income', 'monthly_rent', 'management_fee', 'monthly_debt_payment']].median().round(0)
```
